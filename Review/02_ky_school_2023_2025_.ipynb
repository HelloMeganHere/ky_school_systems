Libraries and Parameters

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn

seaborn.set_theme(style='whitegrid')
```

## KY School Data (2023-2025)

```python
enrollment_2023_2024_df = pd.read_csv("../Dataset/KYRC24_OVW_Student_Enrollment.csv")
enrollment_2024_2025_df = pd.read_csv("../Dataset/KYRC25_OVW_Student_Enrollment.csv")

absenteeism_2023_2024_df = pd.read_csv("../Dataset/KYRC24_OVW_Chronic_Absenteeism.csv")
absenteeism_2024_2025_df = pd.read_csv("../Dataset/KYRC25_OVW_Chronic_Absenteeism.csv")

dropout_2023_2024_df = pd.read_csv("../Dataset/KYRC24_OVW_Dropout_Rate.csv")
dropout_2024_2025_df = pd.read_csv("../Dataset/KYRC25_OVW_Dropout_Rate.csv")


enrollment_df = pd.concat([enrollment_2023_2024_df, enrollment_2024_2025_df], ignore_index=True)
chronic_absenteeism_df = pd.concat([absenteeism_2023_2024_df, absenteeism_2024_2025_df], ignore_index=True)
dropout_rate_df = pd.concat([dropout_2023_2024_df, dropout_2024_2025_df], ignore_index=True)
```
